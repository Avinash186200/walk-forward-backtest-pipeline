# Strategy #

Here are the rules For Swing strategy

1. long entry: Buy when the price crosses above the 50 day EMA and the 100 EMA & when MACD is positive
2. exit: sell when the price crosses below the 50 EMA or the 100 EMA.
3. Short Entry: sell when the price crosses below the 50 EMA and the 100 EMA and when the MACD is negative. 
4. short exit : buy when the price crosses above the 50day EMA or the 100day EMA

 Very simple. And very consistent.
